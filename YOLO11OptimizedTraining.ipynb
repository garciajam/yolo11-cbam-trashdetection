```python
import warnings
import os
warnings.filterwarnings('ignore')
from ultralytics import YOLO

if __name__ == '__main__':
    custom_project_folder = r"/workspace/results/" 
    run_name = "YOLO11m-Optimized"
    checkpoint_path = os.path.join(custom_project_folder, run_name, "weights", "last.pt")

    if os.path.exists(checkpoint_path):
        print(f"Found existing checkpoint. Resuming training from {checkpoint_path}...")
        model = YOLO(checkpoint_path)
        model.train(resume=True)

    else:
        print(f"No checkpoint found. Starting new training. Files will save to: {os.path.join(custom_project_folder, run_name)}")
        model = YOLO('yolo11m-CBAM.yaml')

        model.train(
            data=r"/workspace/PHTrashDataset/data.yaml",
            imgsz=960,
            epochs=300,
            batch=16,
            workers=16,
            device='0',
            optimizer='auto',
            project=custom_project_folder, 
            name=run_name,        
            single_cls=False,
            cache=True,
            amp=True,
            patience=50,
        )
```
